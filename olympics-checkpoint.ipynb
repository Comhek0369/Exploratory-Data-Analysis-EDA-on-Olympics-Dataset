```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: Set style
sns.set(style="whitegrid")
%matplotlib inline
```

```python
df = pd.read_csv("dataset_olympics.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace = True)
```

```python
df.duplicated().sum()
```

```python
sns.countplot(x='Sex', data=df)
plt.title('Distribution by Gender')
plt.show()
```

```python
sns.histplot(data = df, x ="Age", bins = 10, kde =True)
plt.title('Age Distribution')
plt.show()
```

```python
sns.histplot(data = df, x ="Height", bins = 20, kde =True)
plt.title('Height Distribution')
plt.show()
```

```python
sns.histplot(data = df, x ="Weight", bins = 20, kde =True)
plt.title('Weight Distribution')
plt.show()
```

```python
sns.countplot(x='Medal', data=df)
plt.title(' Medal Distribution')
plt.show()
```

```python
sns.countplot(data= df, x = "Year", hue = "Medal")
plt.title ("Medal Distribution Over The Years")
plt.xticks(rotation = 45)
plt.show()
```

```python
year_avg_age = df.groupby("Year")["Age"].mean()
print(year_avg_age)
```

```python
sport_median_height =  df.groupby("Sport")["Height"].median()
print(sport_median_height.max())
```
